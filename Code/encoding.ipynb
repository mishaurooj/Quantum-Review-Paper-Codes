```python
# Importing necessary libraries
from qiskit import QuantumCircuit, Aer, execute
import matplotlib.pyplot as plt
import numpy as np
from qiskit.visualization import plot_bloch_multivector

# 1. Amplitude Encoding (Before and After)
qc_amplitude = QuantumCircuit(3)

# Apply Hadamard gate to create superposition
qc_amplitude.h(0)
qc_amplitude.h(1)
qc_amplitude.h(2)

# Plot the Dynamic Decoupling Circuit
fig = plt.figure(figsize=(2, 2))
ax = fig.add_subplot()
ax.text(1, 1, 'Before Encoding Circuit', size=14)
qc_amplitude.draw(output='mpl', ax=ax)
plt.show()

# Draw before encoding - Bloch sphere visualization
simulator = Aer.get_backend('statevector_simulator')
result_before = execute(qc_amplitude, simulator).result()
statevector_before = result_before.get_statevector()

# Plot Bloch Sphere before encoding
plot_bloch_multivector(statevector_before,  title='Bloch Sphere before encoding')

```

```python
# Importing necessary libraries
from qiskit import QuantumCircuit, Aer, execute
import numpy as np
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt

# Define the function for Chebyshev encoding
def chebyshev_encoding(qc, qubit, degree, data):
    """
    Apply Chebyshev encoding to a specific qubit in the quantum circuit.
    :param qc: Quantum circuit
    :param qubit: qubit to apply encoding to
    :param degree: degree of the Chebyshev polynomial (defines the complexity of encoding)
    :param data: classical data to encode
    """
    for i in range(degree):
        # Chebyshev encoding typically involves applying rotations. Here we use rotations that approximate the Chebyshev polynomials.
        # These rotations are placeholders and can be adjusted to represent specific Chebyshev encoding.
        qc.rx(data[i % len(data)] * (i + 1), qubit)  # Rotate qubit based on Chebyshev approximation
        qc.ry(data[i % len(data)] * (i + 1), qubit)
        qc.rz(data[i % len(data)] * (i + 1), qubit)

# Create a quantum circuit with 3 qubits
qc_chebyshev = QuantumCircuit(3)

# Example classical data to encode using Chebyshev encoding
data = [0.3, 0.6, 0.9]  # Example data

# Apply Hadamard gate to create superposition for the first qubit
qc_chebyshev.h(0)

# Apply Chebyshev encoding to qubit 2 (you can modify this for different qubits)
degree = 3  # Degree of the Chebyshev polynomial encoding
chebyshev_encoding(qc_chebyshev, 2, degree, data)

# Draw the Chebyshev Encoding Circuit (Before encoding)
fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot()
ax.text(1, 1, 'Chebyshev Encoding Circuit', size=14)
qc_chebyshev.draw(output='mpl', ax=ax)
plt.show()

# Simulate the quantum circuit before encoding (Quantum state)
simulator = Aer.get_backend('statevector_simulator')
result_before = execute(qc_chebyshev, simulator).result()
statevector_before = result_before.get_statevector()

# Plot Bloch Sphere before encoding
plot_bloch_multivector(statevector_before, title='Bloch Sphere Before Chebyshev Encoding')

# Simulate the quantum circuit after encoding (Quantum state)
result_after = execute(qc_chebyshev, simulator).result()
statevector_after = result_after.get_statevector()

# Plot Bloch Sphere after encoding
plot_bloch_multivector(statevector_after, title='Chebyshev Encoding Bloch Sphere')

```

```python
# Importing necessary libraries
from qiskit import QuantumCircuit, Aer, execute
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt
import numpy as np

# Define the function for Fourier encoding (Using Quantum Fourier Transform)
def fourier_encoding(qc, qubits):
    """
    Apply Fourier encoding to a set of qubits in the quantum circuit.
    :param qc: Quantum circuit
    :param qubits: List of qubits to apply Fourier encoding
    """
    # Apply Quantum Fourier Transform (QFT) to the qubits
    qc.h(qubits[0])  # Apply Hadamard gate to the first qubit
    for qubit in qubits[1:]:
        qc.h(qubit)  # Apply Hadamard gate to remaining qubits
    
    # Apply controlled rotations to simulate the Fourier Transform
    for i in range(len(qubits)):
        for j in range(i + 1, len(qubits)):
            qc.cp(np.pi / 2**(j - i), qubits[j], qubits[i])  # Controlled phase gates

# Create a quantum circuit with 3 qubits
qc_fourier = QuantumCircuit(3)

# Example classical data to encode (could be any data, for simplicity, just some phases here)
data = [0.1, 0.3, 0.5]  # Example data for Fourier encoding

# Apply Fourier encoding to qubits
fourier_encoding(qc_fourier, [0, 1, 2])

# Draw the Fourier Encoding Circuit
fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot()
ax.text(1, 1, 'Fourier Encoding Circuit', size=14)
qc_fourier.draw(output='mpl', ax=ax)
plt.show()

# Simulate the quantum circuit before and after applying Fourier encoding (Quantum state)
simulator = Aer.get_backend('statevector_simulator')
result_before = execute(qc_fourier, simulator).result()
statevector_before = result_before.get_statevector()

# Plot Bloch Sphere before encoding (In this case, before QFT)
plot_bloch_multivector(statevector_before, title='Bloch Sphere Before Fourier Encoding')

# Simulate the quantum circuit after Fourier encoding (Quantum state)
result_after = execute(qc_fourier, simulator).result()
statevector_after = result_after.get_statevector()

# Plot Bloch Sphere after encoding (Fourier Transform applied)
plot_bloch_multivector(statevector_after, title='Fourier Encoding Bloch Sphere')
```

```python
# Importing necessary libraries
from qiskit import QuantumCircuit, Aer, execute
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt
import numpy as np

# Define the function for Shor's Encoding (9 qubits encoding for 1 logical qubit)
def shors_encoding(qc, qubits):
    """
    Apply Shor's encoding to a qubit (Encode 1 logical qubit into 9 physical qubits).
    :param qc: Quantum circuit
    :param qubits: List of qubits to apply Shor's encoding
    """
    # Encode 3 classical bits into 3 qubits (each classical bit gets 3 physical qubits)
    # Encoding for bit 0 -> Apply X gates to flip qubits if necessary
    qc.cx(qubits[0], qubits[1])
    qc.cx(qubits[0], qubits[2])

    # Encoding for bit 1 -> Apply X gates to flip qubits if necessary
    qc.cx(qubits[3], qubits[4])
    qc.cx(qubits[3], qubits[5])

    # Encoding for bit 2 -> Apply X gates to flip qubits if necessary
    qc.cx(qubits[6], qubits[7])
    qc.cx(qubits[6], qubits[8])

# Create a quantum circuit with 9 qubits for Shor's encoding
qc_shors = QuantumCircuit(9)

# Prepare the qubit in state |0>
# Encode using Shor's code (1 logical qubit encoded into 9 qubits)
shors_encoding(qc_shors, [0, 1, 2, 3, 4, 5, 6, 7, 8])

# Draw the Shor's Encoding Circuit
fig = plt.figure(figsize=(7, 7))
ax = fig.add_subplot()
ax.text(1, 1, 'Shor\'s Encoding Circuit', size=14)
qc_shors.draw(output='mpl', ax=ax)
plt.show()

# Simulate the quantum circuit before and after applying Shor's encoding (Quantum state)
simulator = Aer.get_backend('statevector_simulator')
result_before = execute(qc_shors, simulator).result()
statevector_before = result_before.get_statevector()

# Plot Bloch Sphere before encoding
plot_bloch_multivector(statevector_before, title='Shor\'s Encoding Bloch Sphere')
```
