<p style="font-size:20px; margin-top:12px; margin-left:10px;">
Quantum Information Encoding (QIE) schemes are designed to improve information transmission over noisy quantum channels. These methods combine principles from classical information theory with quantum mechanics to maximize reliability and capacity.
</p>

<!-- 1 -->
<div style="background-color:#ffe0e0; color:#990000; padding:10px; font-weight:bold; font-size:20px; border-radius:8px;">1. Entanglement-Assisted Encoding</div>
<p style="margin-left:20px; font-size:20px;"><b style="color:#cc0000;">Motivation:</b> Uses pre-shared entangled qubits between sender and receiver to boost communication efficiency.</p>
<p style="margin-left:20px; color:#006400; font-size:20px;">✓ Advantage: Increases quantum channel capacity; enables quantum superdense coding.</p>
<p style="margin-left:20px; color:#8b0000; font-size:20px;">✗ Limitation: Requires reliable entanglement distribution and synchronization.</p>


```python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.compiler import transpile
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Entanglement-Assisted Encoding (2-qubit example)
qc_eae = QuantumCircuit(2, 2)
# Create a Bell pair:
qc_eae.h(0)
qc_eae.cx(0, 1)
# Encode information by applying a rotation on qubit 0 (e.g., Ry rotation)
qc_eae.ry(0.8, 0)
# Measure both qubits
qc_eae.measure([0, 1], [0, 1])


# Simulate on qasm simulator
backend = Aer.get_backend('qasm_simulator')
tqc_eae = transpile(qc_eae, backend)
job_eae = backend.run(tqc_eae, shots=1024)
counts_eae = job_eae.result().get_counts()


```

```python
# Plot histogram in a single (distinct) color (e.g. "orange")
plot_histogram(counts_eae, color='orange', title="Entanglement-Assisted Measurement Histogram", figsize=(8,5))
```

```python
# Draw the colored circuit diagram
qc_eae.draw(output='mpl')

```

<!-- 2 -->
<div style="background-color:#e0ffe0; color:#006600; padding:10px; font-weight:bold; font-size:20px; border-radius:8px;">2. Quantum Polar Encoding</div>
<p style="margin-left:20px; font-size:20px;"><b style="color:#cc0000;">Motivation:</b> Extends classical polar codes to quantum channels to approach capacity limits.</p>
<p style="margin-left:20px; color:#006400; font-size:20px;">✓ Advantage: Theoretically achieves quantum channel capacity; efficient for symmetric channels.</p>
<p style="margin-left:20px; color:#8b0000; font-size:20px;">✗ Limitation: Complex construction and decoding for general channels.</p>


```python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.compiler import transpile
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Quantum Polar Encoding (2-qubit conceptual example)
qc_polar = QuantumCircuit(2, 2)
# Prepare qubit 0 in superposition (information splitting)
qc_polar.h(0)
# Polarization transform: combine information from both qubits
qc_polar.cx(0, 1)
# (Optional additional rotations or controlled operations could be added.)
qc_polar.measure([0, 1], [0, 1])



# Simulate on qasm simulator
backend = Aer.get_backend('qasm_simulator')
tqc_polar = transpile(qc_polar, backend)
job_polar = backend.run(tqc_polar, shots=1024)
counts_polar = job_polar.result().get_counts()

```

```python
# Draw the colored circuit diagram
qc_polar.draw(output='mpl')
```

```python
# Plot histogram in a single color (e.g. "teal")
plot_histogram(counts_polar, color='teal', title="Quantum Polar Encoding Histogram", figsize=(8,5))
```

<!-- 3 -->
<div style="background-color:#e6f0ff; color:#003366; padding:10px; font-weight:bold; font-size:20px; border-radius:8px;">3. Quantum LDPC (Low-Density Parity-Check) Encoding</div>
<p style="margin-left:20px; font-size:20px;"><b style="color:#cc0000;">Motivation:</b> Applies sparse parity-check matrices to encode quantum data with low overhead.</p>
<p style="margin-left:20px; color:#006400; font-size:20px;">✓ Advantage: Low complexity, scalable, and suitable for near-term quantum devices.</p>
<p style="margin-left:20px; color:#8b0000; font-size:20px;">✗ Limitation: Requires careful code construction to preserve quantum error correction constraints.</p>


```python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.compiler import transpile
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Quantum LDPC Encoding (Conceptual 5-qubit example)
qc_ldpc = QuantumCircuit(5, 5)
# Prepare one qubit in superposition as information carrier:
qc_ldpc.h(0)
# Sparse entanglement connections (low density)
qc_ldpc.cx(0, 1)
qc_ldpc.cx(0, 2)
qc_ldpc.cx(1, 3)
qc_ldpc.cx(2, 4)
# (In a real LDPC code, the CNOT connectivity would be chosen to reflect a specific sparse parity check matrix.)
qc_ldpc.measure(range(5), range(5))


# Simulate on qasm simulator
backend = Aer.get_backend('qasm_simulator')
tqc_ldpc = transpile(qc_ldpc, backend)
job_ldpc = backend.run(tqc_ldpc, shots=1024)
counts_ldpc = job_ldpc.result().get_counts()
```

```python

# Draw the colored circuit diagram
qc_ldpc.draw(output='mpl')
```

```python
# Plot histogram in a single color (e.g., "purple")
plot_histogram(counts_ldpc, color='purple', title="Quantum LDPC Encoding Histogram", figsize=(8,5))
```

<div style="background-color:#fff0cc; color:#805500; padding:10px; font-weight:bold; font-size:20px; border-radius:8px;">4. Stabilizer Code Encoding</div>
<p style="margin-left:20px; font-size:20px;"><b style="color:#cc0000;">Motivation:</b> General framework to define quantum error-correcting codes using Pauli group stabilizers.</p>
<p style="margin-left:20px; color:#006400; font-size:20px;">✓ Advantage: Basis for most modern QEC codes (e.g., Steane, surface codes).</p>
<p style="margin-left:20px; color:#8b0000; font-size:20px;">✗ Limitation: Requires ancilla qubits and precise syndrome measurements.</p>

```python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.compiler import transpile
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Stabilizer Code Encoding (Conceptual 3-qubit GHZ state)
qc_stab = QuantumCircuit(3, 3)
# Create a GHZ state
qc_stab.h(0)
qc_stab.cx(0, 1)
qc_stab.cx(0, 2)
qc_stab.measure(range(3), range(3))


# Simulate on qasm simulator
backend = Aer.get_backend('qasm_simulator')
tqc_stab = transpile(qc_stab, backend)
job_stab = backend.run(tqc_stab, shots=1024)
counts_stab = job_stab.result().get_counts()
```

```python
# Draw the colored circuit diagram
qc_stab.draw(output='mpl')
```

```python
# Plot histogram in a single color (e.g., "magenta")
plot_histogram(counts_stab, color='magenta', title="Stabilizer Code Encoding Histogram", figsize=(8,5))
```
